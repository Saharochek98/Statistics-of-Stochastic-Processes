```python
import numpy as np
from matplotlib import pyplot as plt
```

Task 1. Simulate a path of the Wiener process. Number of end-points of the grid including $T$ is $N=100$, length of the interval $[0, T]$ in time units is 1, time increment is $\Delta = T/N$. Plot the path.

```python
def WienerRandomValue(t):
    return np.sqrt(t) * np.random.normal(0, 1)


def WienerProcess(N, T):
    t_data = np.linspace(0, T, N + 1)
    w_data = np.zeros(N + 1)
    dt = T/N
    for i in range(1, N + 1):
        w_data[i] = w_data[i - 1] + WienerRandomValue(dt)
    return t_data, w_data

N = 100
T = 1
t_data, w_data = WienerProcess(N, T)

plt.plot(t_data, w_data)
plt.xlabel("t")
plt.ylabel("W(t)")
plt.xlim(0.0, T)
plt.title("Wiener process")
plt.grid(True)
plt.show()
```

Task 2. Using the random walk algorithm simulate three paths of the Wiener process as the limit of a random walk for different $n=10, 100, 1000$. Plot the paths in one figure, add the legend.

```python
def RandomValueP(p):
    assert p > 0.0 and p < 1.0, '0 < p < 1'

    rand = np.random.uniform(0, 1)
    if rand <= p:
        return 1
    else:
        return -1

def RandomWalk(p, N):
    t_data = np.linspace(0, 1, N + 1)
    w_data = np.zeros(N + 1)
    for i in range(1, N + 1):
        w_data[i] = w_data[i - 1] * np.sqrt(i / (i + 1)) + RandomValueP(p) / np.sqrt(i + 1)
    return t_data, w_data

p = 0.5
N = [10, 100, 1000]
for k in N:
    t_data, w_data = RandomWalk(p, k)
    plt.plot(t_data, w_data, label = f"W(t), N = {k}")

plt.legend(loc = 'best', fancybox = True, shadow = True)
plt.xlim(0.0, 1.0)
plt.xlabel("t")
plt.ylabel("W(t)")
plt.title("Wiener process")
plt.grid(True)
plt.show()
```

Task 3. Using the the Karhunen-Loeve expansion simulate three paths of the Wiener process with
$n = 10, 50, 100$ terms. Plot the paths in one figure, add legends.

```python
def phi(k, t, T):
    return 2 * np.sqrt(2 * T) / np.pi / (2 * k + 1) * np.sin((2 * k + 1) * np.pi * t / (2 * T))

def KarhunenLoeveApprox(T, Nt, Nz):
    z_data = np.random.normal(0, 1, size = Nz)
    t_data = np.linspace(0, T, Nt + 1)
    k = np.arange(0, len(z_data)) + 1
    w_data = np.zeros(Nt + 1)
    for i in range(0, Nt + 1):
        phi_data = phi(k, t_data[i], T)
        w_data[i] = np.dot(z_data, phi_data)
    return t_data, w_data

T = 1
Nt = 100
Nz = [10, 50, 100]

for k in Nz:
    t_data, w_data = KarhunenLoeveApprox(T, Nt, k)
    plt.plot(t_data, w_data, label = f'N = {k}')

plt.legend(loc = 'upper left', fancybox = True, shadow = True)
plt.xlim(0.0, 1.0)
plt.xlabel("t")
plt.ylabel("W(t)")
plt.title("Karhunen-Loeve approximation of the Wiener process")
plt.grid(True)
plt.show()
```

Task 4. Plot a trajectory of the geometric Brownian motion obtained from the simulation of the path of the Wiener process, $r=1$ (interest rate) and $\sigma = 0.5$ (volatility).

```python
def GeometricBrownianMotion(S0, r, sigma, T):
    Nt = 1000
    t_data = np.linspace(0, T, Nt + 1)
    s_data = np.zeros(Nt + 1)
    s_data[0] = S0
    dt = T / Nt
    for i in range(1, Nt + 1):
        s_data[i] = s_data[i-1] * np.exp((r - sigma**2 / 2)* dt + sigma * WienerRandomValue(dt))
    return t_data, s_data

S0 = 10
r = 1
sigma = 0.5
T = 1

t_data, s_data = GeometricBrownianMotion(S0, r, sigma, T)
plt.plot(t_data, s_data)
plt.xlim(0.0, 1.0)
plt.xlabel("t")
plt.ylabel("S(t)")
plt.title("Geometric Brownian motion")
plt.grid(True)
plt.show()
```

Task 5. Plot a trajectory of the the Brownian bridge starting at $x$ at time $0$ and terminating its run at $y =-1$ at time $T$ obtained from the simulation of the path of the Wiener process. Add markers for start and end points.

```python
def BrownianBridge(x, y, t0, T):
    Nt = 1000
    t_data, w_data = WienerProcess(Nt, T)
    dt = T/Nt
    bb_data = np.zeros(Nt + 1)
    for i in range(0, Nt + 1):
        bb_data[i] = x + w_data[i] - (t_data[i] - t0) / (T - t0) * (w_data[(Nt - 1)-t0] - y + x)
    return t_data, bb_data

x = 1
y = -1
t0 = 0
T = 5
t_data, bb_data = BrownianBridge(x, y, t0, T)

plt.plot(t_data, bb_data, label = f"BB(t), W({t0}) = {x}, W({T}) = {y}")
plt.scatter ( t0, x, color = "red", s = 50)
plt.scatter ( T, y, color = "red", s = 50)
plt.legend(loc = 'best', fancybox = True, shadow = True)
plt.xlabel("t")
plt.ylabel("BB(t)")
plt.title("Brownian bridge")
plt.grid(True)
plt.show()
```

$$
X_n = \sum_{k=1}^n \xi_k, \quad \xi_k =
\begin{cases}
1, &\text{with probability $\frac12$}; \\
-1, &\text{with probability $\frac12$}.
\end{cases}
$$
$$
\mathbb{E}\xi_k = 0, \quad \mathrm{Var}\, \xi_k = 1.
$$

Plot $(X_{t_i})_{t_i = \frac{i}{n}}$ so that $t_i$ takes values in $[0,\, 1]$.

```python
def RandomWalk(N, p=[0.5, 0.5]):
    xi = np.random.choice([-1, 1], size=N, p=p)

    X = np.cumsum(xi)
    X = np.insert(X, 0, 0)
    t = np.linspace(0, 1, N + 1)

    plt.figure(figsize=(10, 5))
    plt.plot(t, X, marker='o', linestyle='-')
    plt.title('Random Walk')
    plt.xlabel('t')
    plt.ylabel('$X_n$')
    plt.grid()
    plt.show()
```

Exercise 2. Let $H = (H_n)_{n \geq 1}$ be the player's bet at the $n$-th step, i.e. his winnings at this step are equal to $H_n (X_n - X_{n-1}) = H_n \Delta X_n$.

Plot an accumulated winnings of player $Y_n$ at step $n \geq 1$.


```python
def random_walk_with_bets(n, H):
    xi = np.random.choice([-1, 1], size=n, p=[0.5, 0.5])

    X = np.cumsum(xi)

    X = np.insert(X, 0, 0)

    Y = np.zeros(n + 1)
    for i in range(1, n + 1):
        Y[i] = Y[i - 1] + H * (X[i] - X[i - 1])

    t = np.linspace(0, 1, n + 1)

    plt.figure(figsize=(10, 5))
    plt.plot(t, Y, marker='o', linestyle='-', color='orange')
    plt.title('Accumulated Winnings of the Player')
    plt.xlabel('n')
    plt.ylabel('$Y_n$')
    plt.grid()
    plt.show()

random_walk_with_bets(100, H=1)
```

**Task 1. Quadratic Variation (Section 1.1.2)**

For a Wiener process trajectory with $T=1$ and varying numbers of partition intervals $N \in \{10, 100, 1000, 10000\}$:

1. Calculate the sum of squared increments:
$$ S_n = \sum_{k=0}^{n-1} |W(t_{k+1}) - W(t_k)|^2 $$

2. Plot the dependence of $S_n$ on the number of partition intervals (bar chart).

3. Draw conclusions.

```python
def quadratic_variation(w):
    increments = np.diff(w)
    S = np.sum(increments**2)
    return S

N = [10, 100, 1000, 10000]
T = 1
np.random.seed(42)
S_values = []

plt.figure(figsize=(10, 8))
results = {}

for k in N:
    t_data, w_data = WienerProcess(k, T)
    S = quadratic_variation(w_data)
    S_values.append(S)
    print(f"N = {k} ; S_n = {S:.3f}")

plt.figure(figsize=(10, 6))
bars = plt.bar(range(len(N)), S_values, tick_label=N, color='skyblue', edgecolor='navy', alpha=0.7)

for i, (bar, val) in enumerate(zip(bars, S_values)):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.02,
             f'{val:.3f}', ha='center', va='bottom', fontsize=10)

plt.title('Quadratic Variation vs. Number of Intervals (N)', fontsize=12)
plt.xlabel('Number of Intervals (N)', fontsize=10)
plt.ylabel('Quadratic Variation ($S_n$)', fontsize=10)
plt.xticks(range(len(N)), N)
plt.grid(True, alpha=0.3, axis='y')
plt.legend()

plt.tight_layout()
plt.show()
```

### Conclusion:

As the number of partition intervals $N$ increases, the quadratic variation $S_n$ converges to 1 (for $T=1$). This is a fundamental property of the Wiener process.


**Task 2. Empirical Covariance Function (Section 1.1.3)**

For a Wiener process, the theoretical covariance between states at times $s$ and $t$ is $\min(s,t)$.

1. Generate an ensemble of 500 Wiener process trajectories.

2. Compute the sample covariance between points $t=0.5$ and $t=T$, and compare the result with the theoretical value of $0.5$.

3. Draw conclusions.

```python
N_traj= 500
N = 1000
T = 1
t1 = 0.5
t2 = T

np.random.seed(42)
W_t1_values = []
W_t2_values = []

for i in range(N_traj):
    t, W = WienerProcess(N, T)

    idx_t1 = int(t1 * N)
    idx_t2 = -1

    W_t1_values.append(W[idx_t1])
    W_t2_values.append(W[idx_t2])

W_t1_values = np.array(W_t1_values)
W_t2_values = np.array(W_t2_values)

theoretical_cov = min(t1, t2)
empirical_cov = np.cov(W_t1_values, W_t2_values)[0, 1]
empirical_corr = np.corrcoef(W_t1_values, W_t2_values)[0, 1]

print(f"Theoretical covariance Cov(W(0.5), W(1)) = min(0.5, 1) = {theoretical_cov}")
print(f"Empirical covariance (based on {N_traj} trajectories): {empirical_cov:.3f}")
print(f"Relative error: {abs(empirical_cov - theoretical_cov)/theoretical_cov*100:.3f}%")
```

### Conclusion:

The empirical covariance, estimated from an ensemble of trajectories, is in close agreement with the theoretical value $\min(0.5, 1) = 0.5$.

This result verifies the fundamental covariance property of the Wiener process:
$$ \text{Cov}(W(s), W(t)) = \min(s, t) $$

Furthermore, increasing the number of trajectories in the ensemble leads to a more accurate estimate, consistent with the Law of Large Numbers.

**Task 3: Testing the Martingale Property (Section 1.1.5)**

**Objective:** Verify the martingale property of the Wiener process through conditional expectation analysis.

1. Generate an ensemble of 500 Wiener process trajectories.

2. Select a fixed time point $s = 0.4$.

3. For all generated trajectories that pass through a specific narrow range of values at time $s$, compute the arithmetic mean of their values at the terminal time $T$.

4. Compare the computed arithmetic mean with the process value at time $s$.

5. Draw conclusions. Analyze how the range width affects the result.

```python
N_traj = 500
N = 1000
T = 1
s = 0.4
delta = 0.2

np.random.seed(42)

all_trajectories = []
W_s_values = []
W_T_values = []

for i in range(N_traj):
    t, W = WienerProcess(N, T)
    all_trajectories.append((t, W))

    idx_s = int(s * N)
    W_s_values.append(W[idx_s])
    W_T_values.append(W[-1])

W_s_values = np.array(W_s_values)
W_T_values = np.array(W_T_values)

center = np.mean(W_s_values)
lower_bound = center - delta/2
upper_bound = center + delta/2

mask = (W_s_values >= lower_bound) & (W_s_values <= upper_bound)
selected_indices = np.where(mask)[0]
n_selected = len(selected_indices)

print(f"\nRange width Δ = {delta}")
print(f"Range boundaries: [{lower_bound:.3f}, {upper_bound:.3f}]")
print(f"Number of trajectories in range: {n_selected} out of {N_traj}")

if n_selected > 0:
    mean_W_T = np.mean(W_T_values[mask])
    mean_W_s = np.mean(W_s_values[mask])


    plt.figure(figsize=(10, 6))

    for i, (t_i, W_i) in enumerate(all_trajectories):
        if i < 100:
            if mask[i]:
                plt.plot(t_i, W_i, 'r-', alpha=0.7, linewidth=1.0, label='In range' if i == selected_indices[0] else "")
            else:
                plt.plot(t_i, W_i, 'b-', alpha=0.1, linewidth=0.5)


    plt.axvline(x=s, color='green', linestyle='--', linewidth=2, label=f'Moment s = {s}')
    plt.axhline(y=upper_bound, color='orange', linestyle=':', linewidth=2, alpha=0.8)
    plt.fill_between([0, T], lower_bound, upper_bound, color='yellow', alpha=0.2, label=f'Range')

    plt.plot(s, mean_W_s, 'go', markersize=12, label=f'Mean W({s}) = {mean_W_s:.3f}')
    plt.plot(T, mean_W_T, 'ro', markersize=12, label=f'Mean W(T) = {mean_W_T:.3f}')

    plt.plot([s, T], [mean_W_s, mean_W_T], 'k--', linewidth=2, alpha=0.7, label=f'Difference: {mean_W_T - mean_W_s:.4f}')

    plt.xlabel('t', fontsize=12)
    plt.ylabel('W(t)', fontsize=12)
    plt.title(f'Wiener Process Trajectories\nRange Δ = {delta} at time s = {s} ({n_selected} trajectories in range)', fontsize=14)
    plt.grid(True, alpha=0.3)
    plt.legend(loc='upper left', fontsize=10)
    plt.xlim(0, T)

    plt.tight_layout()
    plt.show()


    print(f"Mean of W({s}) within the range: {mean_W_s:.4f}")
    print(f"Mean of W(T) for selected trajectories: {mean_W_T:.4f}")
    print(f"Difference (deviation from martingale property): {mean_W_T - mean_W_s:.4f}")

    if abs(mean_W_T - mean_W_s) < 0.05:
        print("Martingale property holds (deviation is within the statistical error margin)")
    else:
        print("Deviation from the martingale property exceeds the expected statistical error margin")
```

### Conclusion:

**Verification of the Martingale Property:**
For a Wiener process, the martingale property implies:
$$ \mathbb{E}[W(T) \mid \mathcal{F}_s] = W(s) $$
where $\mathcal{F}_s$ represents the information available up to time $s$ (the natural filtration).

In our experiment, we empirically verified that:
$$ \mathbb{E}[W(T) \mid W(s) \in [a, b]] \approx \text{mean of } W(s) \text{ within this range.} $$

**Task 4: Analysis of Independent Increments (Section 1.2)**

**Objective:** Empirically verify the independent increments property of the Wiener process.

1. Select two non-overlapping time intervals $[t_1, t_2]$ and $[t_3, t_4]$.

2. Generate $M$ trajectories and compute the increments:
$$ \Delta W_1 = W(t_2) - W(t_1) \quad \text{and} \quad \Delta W_2 = W(t_4) - W(t_3) $$

3. Plot a scatter plot of $(\Delta W_1, \Delta W_2)$ and compute the Pearson correlation coefficient.

4. Compare the resulting correlation coefficient with zero and draw conclusions.

```python
M = 500
N = 1000
T = 1

t1, t2 = 0.2, 0.4
t3, t4 = 0.6, 0.8

np.random.seed(42)

delta_W1 = []
delta_W2 = []

for i in range(M):
    t, W = WienerProcess(N, T)

    idx_t1 = int(t1 * N)
    idx_t2 = int(t2 * N)
    idx_t3 = int(t3 * N)
    idx_t4 = int(t4 * N)

    dW1 = W[idx_t2] - W[idx_t1]
    dW2 = W[idx_t4] - W[idx_t3]

    delta_W1.append(dW1)
    delta_W2.append(dW2)

delta_W1 = np.array(delta_W1)
delta_W2 = np.array(delta_W2)

correlation = np.corrcoef(delta_W1, delta_W2)

plt.figure(figsize=(8, 6))
plt.scatter(delta_W1, delta_W2, alpha=0.5, s=30, c='blue', edgecolors='black', linewidth=0.5)
plt.xlabel(f'ΔW₁ = W({t2}) - W({t1})', fontsize=12)
plt.ylabel(f'ΔW₂ = W({t4}) - W({t3})', fontsize=12)
plt.title(f'Scatter plot of increments (ΔW₁, ΔW₂)\nTesting independent increments property', fontsize=14)
plt.grid(True, alpha=0.3)
plt.axhline(y=0, color='gray', linestyle='-', linewidth=0.5)
plt.axvline(x=0, color='gray', linestyle='-', linewidth=0.5)
plt.axis('equal')

plt.tight_layout()
plt.show()

print(f"Interval 1: [{t1}, {t2}]")
print(f"Interval 2: [{t3}, {t4}]")
print(correlation)
print(f"Mean of ΔW₁: {np.mean(delta_W1):.4f} (theoretical: 0)")
print(f"Mean of ΔW₂: {np.mean(delta_W2):.4f} (theoretical: 0)")
print(f"Variance of ΔW₁: {np.var(delta_W1):.4f} (theoretical: {t2-t1:.2f})")
print(f"Variance of ΔW₂: {np.var(delta_W2):.4f} (theoretical: {t4-t3:.2f})")
```

### Conclusion:

The correlation is close to zero, which confirms the independent increments property of the Wiener process on non-overlapping time intervals.

**Task 5: Accuracy of Karhunen-Loève Expansion (Section 1.2.3)**

**Objective:** Analyze the convergence of the Karhunen-Loève expansion for approximating Wiener process trajectories.

1. Implement the approximation of a Wiener trajectory using $n \in \{10, 100, 1000\}$ terms.

2. Calculate the mean squared error (MSE) between the direct simulation and the Karhunen-Loève approximation.

3. Plot the error as a function of the number of expansion terms $n$.

4. Analyze the convergence behavior and draw conclusions.

```python
T = 1
Nt = 1000
n_values = [10, 100, 1000, 10000]
errors = []
np.random.seed(42)

t_ref, w_ref = WienerProcess(Nt, T)
plt.figure(figsize=(10, 6))

for n in n_values:
    t_kl, w_kl = KarhunenLoeveApprox(T, Nt, n)
    mse = np.mean((w_ref - w_kl) ** 2)
    errors.append(mse)
    plt.plot(t_kl, w_kl, '--', label=f'KL n={n} (MSE={mse:.4f})')

plt.plot(t_ref, w_ref, 'k-', linewidth=2, label='Direct simulation')

plt.xlabel('t')
plt.ylabel('W(t)')
plt.title('Wiener Process Approximation via Karhunen-Loève Expansion')
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 5))
plt.semilogy(n_values, errors, 'bo-', linewidth=2, markersize=8)
plt.xlabel('Number of terms (n)')
plt.ylabel('Mean Squared Error (MSE)')
plt.title('Error vs. Number of Terms')
plt.grid(True)
plt.show()

print("Errors for different n:")
for n, err in zip(n_values, errors):
    print(f"n = {n}: MSE = {err:.6f}")
```

### Conclusion:

The analysis confirms the theoretical convergence of the Karhunen-Loève expansion to the true Wiener process. As the number of terms $n$ increases, the Mean Squared Error (MSE) exhibits a clear decreasing trend, approaching zero.

Furthermore, the results highlight the trade-off between approximation accuracy and model complexity:
- A small number of terms ($n=10$) captures only the low-frequency, general trend of the process.
- A larger number of terms ($n=1000$) is required to accurately reproduce the high-frequency variations (roughness) inherent to Wiener trajectories.
